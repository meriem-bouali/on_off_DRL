```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv("density_02-02-2026.csv")

# Plot each edge
for col in df.columns[1:]:  # skip "Timestep"
    plt.plot(df["Timestep"], df[col], label=col)

plt.xlabel("Timestep")
plt.ylabel("Density (pc/mi/ln)")
plt.title("Edge Density Over Time")
# Move legend outside
plt.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    borderaxespad=0
)
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("density_27-01-2026.csv")

for col in df.columns[1:]:
    plt.scatter(df["Timestep"], df[col], label=col, s=20)

plt.xlabel("Timestep")
plt.ylabel("Density (pc/mi/ln)")
plt.title("Edge Density (Dots)")

# Move legend outside
plt.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    borderaxespad=0
)

plt.grid(True)
plt.tight_layout()
plt.show()
```
