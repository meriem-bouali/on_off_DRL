## metrics_summary in CSV

```python
# Directory containing the CSV files
DATA_DIR = ["test_seed_train","test_seed_test"]

# Map filenames to model names
file_model_map = {
    "DoubleDQN_OnRL.csv":"On-DDQN",
    "DoubleDQN_OffRL4e6.csv":"Off-DDQN-4e6",
    "BCQ_BCQ4e6.csv":"BCQ-4e6",
    "DoubleDQN_OffRL4e6Clean.csv":"Off-DDQN-4e6-Clean",
    "BCQ_BCQ4e6Clean.csv":"BCQ-4e6-Clean",
    "DoubleDQN_OffRL4e6Shuffle.csv":"Off-DDQN-4e6-Shuffle",
    "BCQ_BCQ4e6Shuffle.csv":"BCQ-4e6-Shuffle",
    "DoubleDQN_OffRL4e6CleanShuffle.csv":"Off-DDQN-4e6-Clean-Shuffle",
    "BCQ_BCQ4e6CleanShuffle.csv":"BCQ-4e6-Clean-Shuffle",
    "DoubleDQN_OffRL8e6Clean.csv":"Off-DDQN-8e6-Clean",
    "BCQ_BCQ8e6Clean.csv":"BCQ-8e6-Clean",
}


for DIR in DATA_DIR:
    records = []

    for file_name, model_name in file_model_map.items():
        filepath = os.path.join(HYPER_PARAMS.log_dir, DIR, file_name)

    
        df = pd.read_csv(filepath)

        record = {
            "Model": model_name,

            # Episode total reward
            "Avg. Episode Reward": df["episode_total_reward"].mean(),
            # "min_episode_total_reward": df["episode_total_reward"].min(),
            # "max_episode_total_reward": df["episode_total_reward"].max(),
            
            # Episode length
            # "avg_episode_length": df["episode_length"].mean(), # avg_vehicle_speed is more accurate
            # "min_episode_length": df["episode_length"].min(),
            # "max_episode_length": df["episode_length"].max(),
            
            # Vehicle speed
            "Avg. Veh. Speed": df["episode_avg_speed"].mean(),
            # "min_vehicle_speed": df["episode_avg_speed"].min(),
            # "max_vehicle_speed": df["episode_avg_speed"].max(),

            # Total counts
            # "total_nb_lc": df["episode_nb_lc"].sum(),
            "Avg. Episode Nb. LC": df["episode_nb_lc"].mean(),
            "Total Nb. Min. Gap Viol.": df["episode_nb_mingap_violation"].sum(),
            "Total Nb. Invalid LC": df["episode_nb_invalide_lc"].sum(),
            "Total Nb. Veh. Collisions": df["episode_nb_vehicle_collision"].sum(),
        }

        records.append(record)
        print(f"✓ Processed: {model_name} ({len(df)} episodes)")

    output_path = "./metrics_summary_"+DIR[5:]+".csv"
    

    summary_df = pd.DataFrame(records).set_index("Model")
    summary_df = summary_df.round(2)
    summary_df = summary_df.T
    summary_df.index.name = "Metric"

    summary_df.to_csv(output_path)
```

```python
# All seeds combined
print("\nProcessing: all seeds (combined)")
all_records = []

for file_name, model_name in file_model_map.items():
    dfs = [pd.read_csv(os.path.join(HYPER_PARAMS.log_dir,DIR, file_name)) for DIR in DATA_DIR] 
    df = pd.concat(dfs, ignore_index=True)

    record = {
            "Model": model_name,

            # Episode total reward
            "Avg. Episode Reward": df["episode_total_reward"].mean(),
            # "min_episode_total_reward": df["episode_total_reward"].min(),
            # "max_episode_total_reward": df["episode_total_reward"].max(),

            # Episode length
            # "avg_episode_length": df["episode_length"].mean(), # avg_vehicle_speed is more accurate
            # "min_episode_length": df["episode_length"].min(),
            # "max_episode_length": df["episode_length"].max(),

            # Vehicle speed
            "Avg. Veh. Speed": df["episode_avg_speed"].mean(),
            # "min_vehicle_speed": df["episode_avg_speed"].min(),
            # "max_vehicle_speed": df["episode_avg_speed"].max(),

            # Total counts
            # "total_nb_lc": df["episode_nb_lc"].sum(),
            "Avg. Episode Nb. LC": df["episode_nb_lc"].mean(),
            "Total Nb. Min. Gap Viol.": df["episode_nb_mingap_violation"].sum(),
            "Total Nb. Invalid LC": df["episode_nb_invalide_lc"].sum(),
            "Total Nb. Veh. Collisions": df["episode_nb_vehicle_collision"].sum(),
        }
    all_records.append(record)
    print(f"✓ Processed: {model_name} ({len(df)} episodes)")

summary_df = pd.DataFrame(all_records).set_index("Model")
summary_df = summary_df.round(2).T
summary_df.index.name = "Metric"
summary_df.to_csv("./metrics_summary_all_seeds.csv")
```

```python
import pandas as pd
import os

# Map filenames to model names
file_model_map = {
    "DoubleDQN_OnRL.csv":"On-DDQN",
    "DoubleDQN_OffRL4e6.csv":"Off-DDQN-4e6",
    "BCQ_BCQ4e6.csv":"BCQ-4e6",
    "DoubleDQN_OffRL4e6Clean.csv":"Off-DDQN-4e6-Clean",
    "BCQ_BCQ4e6Clean.csv":"BCQ-4e6-Clean",
    "DoubleDQN_OffRL4e6Shuffle.csv":"Off-DDQN-4e6-Shuffle",
    "BCQ_BCQ4e6Shuffle.csv":"BCQ-4e6-Shuffle",
    "DoubleDQN_OffRL4e6CleanShuffle.csv":"Off-DDQN-4e6-Clean-Shuffle",
    "BCQ_BCQ4e6CleanShuffle.csv":"BCQ-4e6-Clean-Shuffle",
    "DoubleDQN_OffRL8e6Clean.csv":"Off-DDQN-8e6-Clean",
    "BCQ_BCQ8e6Clean.csv":"BCQ-8e6-Clean",
}

# Directory containing the CSV files
DATA_DIR = ["./test_seed_train","./test_seed_test"]

for DIR in DATA_DIR:

    merged = None

    for file_name, model_name in file_model_map.items():

        df = pd.read_csv(os.path.join(HYPER_PARAMS.log_dir,DIR, file_name))  

        df["episode_total_reward"] = df["episode_total_reward"].round(2)

        # KEEP ONLY  seed and episode_total_reward only 
        df = df[["seed", "episode_total_reward"]]

        # Rename column to model name
        df = df.rename(columns={"episode_total_reward": model_name})

        # Merge
        if merged is None:
            merged = df
        else:
            merged = pd.merge(merged, df, on="seed", how="outer")

    # Save final table
    merged.to_csv("./episode_reward_"+DIR[7:]+".csv", index=False)
```

## 95% Confidence Intervals

```python
import pandas as pd
import numpy as np
from scipy.stats import bootstrap

pd.set_option('display.width', 2000)              # increase total line width
pd.set_option('display.expand_frame_repr', False) # disable wrapping into multiple rows

def bootstrap_ci_95(data, file_name):
    data = data.drop(columns=["seed"])

    results = {}

    for col in data.columns:
        values = data[col].dropna().values

        res = bootstrap(
            (values,),
            statistic=np.mean,
            n_resamples=10000,
            confidence_level=0.95,
            method="percentile",  # or "basic", "BCa"
            random_state=0
        )

        results[col] = [
        round(np.mean(values), 2),
        round(res.confidence_interval.high, 2),
        round(res.confidence_interval.low, 2)
    ]

    # Convert to DataFrame
    results_df = pd.DataFrame(
        results,
        index=["Avg. Episode Reward", "Upper 95% CI ", "Lower 95% CI"]
    )

    print(results_df, "\n")

    # save to CSV
    results_df.to_csv(f"95_CI_{file_name}.csv")
```

```python
seed_test = pd.read_csv("episode_reward_seed_test.csv")
seed_train = pd.read_csv("episode_reward_seed_train.csv")
all_seed = pd.concat([seed_test, seed_train], axis=0)
```

```python
bootstrap_ci_95(seed_train, "seed_train")
bootstrap_ci_95(seed_test, "seed_test")
bootstrap_ci_95(all_seed, "all_seed")
```
